```python
#We are importing padas library in our model creation process in order to support all the functions related to data set 
#we also alising padas as pd for code simplification
import pandas as pd
```

```python
#using our pd library we calling read_csv function to call our employee salary data set
#This process is datacollection
empsaldataset=pd.read_csv("Employee_Salary_Data.csv")
empsaldataset
```

```python
#Now we need to split our input and output data
#We have stored our employee salary dataset in empsaldataset variable, using that we will split input and output
input_emp_years=empsaldataset[["YearsofExperience"]]
output_emp_sal=empsaldataset[["Salary"]]
print(input_emp_years)
print(output_emp_sal)
```

```python
#Now we need to split our training and test data
#All Modelcreation functions for MACHINE LEARNING available under sklearn Library
#We importing only train_test_split function for sklearn

from sklearn.model_selection import train_test_split

#calling the function with our input parameters
#these function will return 4 fours values, hence storing using 4 variables
x_train,x_test,y_train,y_test=train_test_split(input_emp_years,output_emp_sal,test_size=0.30,random_state=0)
print(x_train)
print(y_train)
print(x_test)
print(y_test)

```

```python
#Actual model creation step, identifying the value of weight and bias to substitute in formula y=wx+b

from sklearn.linear_model import LinearRegression
regressor=LinearRegression()

#now calling exact function for model creation by passing input and output training data

regressor.fit(x_train,y_train)
```

```python
#to identify the weight of the model creation
weight=regressor.coef_
print ("weight=",weight)

#to identify the bias of the model

bias=regressor.intercept_
print("bias=",bias)
```

```python
#Identifying the evaluation metrics by passing the input test data

y_predicted_output=regressor.predict(x_test)
print("Predicted output",y_predicted_output)
print("Actual output",y_test)
```

```python
#Identifying the score of our model using Squared R2 evalution metric

from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_predicted_output)
print(r_score)
```

```python
#Now we need to save our model using pickle library

import pickle
filename="final_empsalaryprediction_linear_regressionmodel.sav"

#all our model data stored in regressor, we dumping our regressor model into the file by writing in binary format
pickle.dump(regressor,open(filename,'wb'))
```
