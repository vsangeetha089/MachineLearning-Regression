```python
import pandas as pd
```

```python
#datacollection
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
#data preprocessing
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
#split input and output data
input_data=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
print(input_data)
output_data=dataset[['charges']]
print(output_data)
```

```python
#split train and test
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(input_data,output_data,test_size=0.30,random_state=0)
print(x_train)
print(y_train)
print(x_test)
print(y_test)
```

```python
#module creation
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(n_estimators=1000,criterion='poisson')
regressor=regressor.fit(x_train,y_train)
```

```python
y_predict=regressor.predict(x_test)
print(y_predict)
print(y_test)
```

```python
#evaluation metrics
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_predict)
print(r_score)
```
