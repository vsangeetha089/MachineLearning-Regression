```python
import pandas as pd
```

```python
#datacollection
dataset=pd.read_csv("50_Startups.csv")
dataset
```

```python
#preprocessing categorical data
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
#spliting input and output data
input_data=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
output_data=dataset[['Profit']]
print(input_data)
print(output_data)
```

```python
#spliting training and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(input_data,output_data,test_size=0.30,random_state=0)
print(x_train)
print(y_train)
print(x_test)
print(y_test)
```

```python
#standardizing the input values
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
print(x_train)
print(x_test)
```

```python
#Gridsearch process started
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'kernel':['rbf','poly','sigmoid','linear'],'C':[10,100,1000,2000,3000],'gamma':['auto','scale']}
regressor_grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
regressor_grid.fit(x_train,y_train)
```

```python
re=regressor_grid.cv_results_
table=pd.DataFrame.from_dict(re)
table
```

```python
print('The best model is ',regressor_grid.best_params_)
```

```python
print('The R_Score value for best parameter{}'.format(regressor_grid.best_params_))
```

```python
future_prediction=regressor_grid.predict([[12345,5334,654,0,1]])
print(future_prediction)
```
